```python
#Handling Missing values

#1. Remove rows with missing vales
#2. Replace missing values with something like "Unknown"

df["Phone_Number"] = df["Phone_Number"].fillna("Unknown")
```

```python
#Drop values
df.dropna(subset = ["Date_Registered"], inplace = True)

#inplace - without creating another variable

#Missing values
df.isnull().sum()
```

```python
#Replacing missing gender with the mode
df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])

df.isnull().sum()
```

```python
#Gender cleaning

#Unique values
df["Gender"].unique()
```

```python
#Value counts
df["Gender"].value_counts()
```

```python
df["Gender"] = df["Gender"].replace({
    "female": "Female",
    "F": "Female",
    "Female ": "Female",
    "M": "Male",
    "male":"Male",
    "MALE":"Male"
})

#Confirm replacements
df["Gender"].value_counts()
```

```python
#Duplicate values using Unique ID
df[df.duplicated(subset = "Customer_ID", keep = False)]
```

```python
df.drop_duplicates(inplace = True)

#Crosschecking
df[df.duplicated(subset = ["Customer_ID","Date_Registered"], keep = False)]
```

```python
df.drop_duplicates(subset = ["Customer_ID","Date_Registered"],inplace = True)

#Crosschecking
df[df.duplicated(subset = ["Customer_ID","Date_Registered"], keep = False)]
```

```python
df["Monthly_Income_KES"] = pd.to_numeric(df["Monthly_Income_KES"], errors = "coerce")

df["Last_Payment_KES"] = pd.to_numeric(df["Last_Payment_KES"], errors = "coerce")

df.dtypes
```

```python
#If the columns have no errors and are clean use .astype(float/int)
df["Monthly_Income_KES"] = df["Monthly_Income_KES"].astype(float)

df["Last_Payment_KES"] = df["Last_Payment_KES"].astype(float)
 
df.dtypes
```

```python
#Detecting outliers

#Sorting missing values
df["Monthly_Income_KES"] = df["Monthly_Income_KES"].fillna(df["Monthly_Income_KES"].median())

Q1 = df["Monthly_Income_KES"].quantile(0.25)
Q3 = df["Monthly_Income_KES"].quantile(0.75)

print(Q1)
print(Q3)

#Interquatile Range calculations

IQR = Q3 - Q1
lower_bound = Q1 - 1.5*IQR
upper_bound = Q3 + 1.5*IQR

#Display Outliers
outliers = df.loc[(
    df["Monthly_Income_KES"]<lower_bound)|(df["Monthly_Income_KES"]>upper_bound)]

outliers
```

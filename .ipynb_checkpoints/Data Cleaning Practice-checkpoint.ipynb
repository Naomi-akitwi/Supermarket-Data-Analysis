```python
#Inconsistent category labels in Gender, County, Employment_Type, Repayment_Status and Payment_Channel
unique_gender = df["Gender"].unique()
unique_gender
```

```python
df["Gender"] = df["Gender"].replace({
    "female": "Female",
    "F": "Female",
    "Female ": "Female",
    "m": "Male",
    "male":"Male",
    "MALE":"Male",
    "nan":"UNKNOWN"
})

#Confirm replacements
df["Gender"].value_counts()
```

```python
unique_counties = df["County"].unique()
unique_counties
```

```python
df["County"] = df["County"].replace({
    "Nairobi ": "Nairobi",
    "nairobi": "Nairobi",
    "kiambu ": "Kiambu",
    "Kisumu ": "Kisumu"
})

#Confirm replacements
df["County"].value_counts()
```

```python
df["County"] = df["County"] . str.strip().str.title()
df
#Confirm replacements
df["County"].value_counts()
```

```python
#Employment type
unique_employement = df["Employment_Type"].unique()
unique_employement
```

```python
df["Employment_Type"] = df["Employment_Type"].replace({
    "Salaried ": "Salaried",
    "self employed": "Self-employed",
    "SELF-EMPLOYED": "Self-employed",
    "casual worker": "Casual worker"
})

#Confirm replacements
df["Employment_Type"].value_counts()
```

```python
#Repayment status
unique_repayment = df["Repayment_Status"].unique()
unique_repayment
```

```python
df["Repayment_Status"] = df["Repayment_Status"].replace({
    "PAID ": "Paid",
    "paid": "Paid",
    "OVERDUE": "Overdue"
})

#Confirm replacements
df["Repayment_Status"].value_counts()
```

```python
#Payment_Channel
unique_channel = df["Payment_Channel"].unique()
unique_channel
```

```python
df["Payment_Channel"] = df["Payment_Channel"].replace({
    "M PESA": "M-Pesa",
    "mpesa": "M-Pesa",
    "bank transfer": "Bank Transfer"
})

#Confirm replacements
df["Payment_Channel"].value_counts()
```

```python
# loans whose due date is on or before 30 September 2026
matured_loan = genuine_disbursed[genuine_disbursed["Due_Date"] <= "2026-09-30"]

total_matured = len(matured_loan)

fully_repaid = len(matured_loan[matured_loan["Amount_Repaid_KES"] >= matured_loan["Total_Amount_Due_KES"]])
not_repaid = len(matured_loan[matured_loan["Amount_Repaid_KES"] == 0])
partially_repaid = len(matured_loan[(matured_loan["Amount_Repaid_KES"] > 0) & (matured_loan["Amount_Repaid_KES"] < matured_loan["Total_Amount_Due_KES"])])

# Percentages
fully_pct = (fully_repaid / total_matured) * 100
part_pct = (partially_repaid / total_matured) * 100
not_pct = (not_repaid / total_matured) * 100

print(fully_pct)
print(part_pct)
print(not_pct)
```

```python
#Outstanding amount as Total_Amount_Due_KES - Amount_Repaid_KES for disbursed loans
genuine_disbursed["Outstanding_Amount_KES"] = (
    genuine_disbursed["Total_Amount_Due_KES"] - genuine_disbursed["Amount_Repaid_KES"]
).clip(lower=0)

overdue_mask = genuine_disbursed["Due_Date"] <= "2026-09-30"

overdue = genuine_disbursed[overdue_mask]
not_yet_overdue = genuine_disbursed[~overdue_mask]

#Sum up the outstanding amounts
total_outstanding = genuine_disbursed["Outstanding_Amount_KES"].sum()
overdue_outstanding = overdue["Outstanding_Amount_KES"].sum()
not_yet_due_outstanding = not_yet_overdue["Outstanding_Amount_KES"].sum()

print(total_outstanding)
print(overdue_outstanding)
print(not_yet_due_outstanding)
```

```python
#Overdue loan Prevalence by county and payment methods
matured_loan = genuine_disbursed[genuine_disbursed["Due_Date"] <= "2026-09-30"]

#overdue
matured_loan["Is_Overdue"] = matured_loan["Outstanding_Amount_KES"] > 0

#Group by County and Employment Type
prevalence_summary = matured_loan.groupby(["County", "Employment_Type"])["Is_Overdue"].agg(
    Total_Matured_Loans="count",
    Overdue_Loans="sum"
).reset_index()

# Prevalence percentage
prevalence_summary["Overdue_Prevalence_%"] = (
    (prevalence_summary["Overdue_Loans"] / prevalence_summary["Total_Matured_Loans"]) * 100
).round(2)

print(prevalence_summary)
```

```python
#Credit score vs repayment performance
bins = [0, 500, 600, 700, 850]
labels = ["Poor (<500)", "Fair (500-600)", "Good (601-700)", "Excellent (701+)"]

#credit score bands
matured_loan["Credit_Score_Band"] = pd.cut(matured_loan["Credit_Score"], bins=bins, labels=labels)

#total loans vs. overdue loans
score_summary = matured_loan.groupby("Credit_Score_Band")["Is_Overdue"].agg(
    Total_Matured_Loans="count",
    Overdue_Loans="sum"
).reset_index()

#Prevalence percentage for each band
score_summary["Overdue_Prevalence_%"] = (
    (score_summary["Overdue_Loans"] / score_summary["Total_Matured_Loans"]) * 100
).round(2)

print(score_summary)
```

```python
#5 largest outstanding loan balances
sorted_loans = matured_loan.sort_values(by="Outstanding_Amount_KES", ascending=False)

top_5_largest = sorted_loans.head(5)
top_5_largest
```

```python
#Approved versus declined applications, bar chart.
import matplotlib.pyplot as plt

#Applications by status
appl_Status = df["Approval_Status"].value_counts()

#Visualization
appl_Status.plot (kind = "bar", color = ["green"])
plt.title("Approved VS Declined Applications")
plt.xlabel("Application Status")
plt.ylabel("Number of Applications")
plt.show()
```

```python
#Number of approved loans by county, horizontal bar chart
#Approved status by county
approved = df[df["Approval_Status"]=="Approved"]
approved_county = approved["County"].value_counts()

#Visualization
approved_county.plot (kind = "barh", color = ["brown"])
plt.title("Approved Loans by County")
plt.xlabel("Number of Approved Loans")
plt.ylabel("County")
plt.show()
```

```python
#Disbursed principal by loan product, bar chart
product_principal = df.groupby("Loan_Product")["Loan_Amount_KES"].sum()
product_principal

#Visualization
product_principal.plot (kind = "bar", color = ["indigo"])
plt.title("Disbursed Principal by Loan Product")
plt.xlabel("Loan Product")
plt.ylabel("Total Disbursed")
plt.show()
```

```python
#Overdue prevalence by credit-score band, bar chart
hazard_colors = ["Red","Orange","green","blue"]
#Visualization
score_summary.plot (x = "Credit_Score_Band", y = "Overdue_Prevalence_%", kind = "bar", color = hazard_colors)
plt.title("Overdue prevalence by credit-score band")
plt.xlabel("credit-score band")
plt.ylabel("% Overdue prevalence")
plt.show()
```

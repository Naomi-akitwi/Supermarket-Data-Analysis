```python
product = df["Product line"]
units = df["Unit price"]
quantity = df["Quantity"]

totals = list(zip(product,units,quantity))
```

```python
print(totals)
```

```python

product_totals = {}

for p, u, q in totals:
    total1 = u * q
    if p in product_totals:
        product_totals[p] += total1
    else:
        product_totals[p] = total1

# Print the aggregated total for each product
for product, total in product_totals.items():
    print(f"{product} - Ksh. {round(total, 2)}")
```

```python
f_ewallet = totals1.count(("Female","Ewallet"))
f_cash = totals1.count(("Female","Cash"))
f_card = totals1.count(("Female","Credit card"))

m_ewallet = totals1.count(("Male","Ewallet"))
m_cash = totals1.count(("Male","Cash"))
m_card = totals1.count(("Male","Credit card"))

#Female mode
female_max = max(f_ewallet,f_cash,f_card)
if female_max == f_ewallet:
    female_mode = "f_ewallet"
elif female_max == f_cash:
    female_mode = "Cash"
else: 
    female_mode = "Credit card"

#Male mode
male_max = max(m_ewallet,m_cash,m_card)
if male_max == m_ewallet:
    male_mode = "m_ewallet"
elif male_max == m_cash:
    male_mode = "Cash"
else: 
    male_mode = "Credit card"
print (f"Female Mode: {female_mode} (Count:{female_max})")
print (f"Male Mode: {male_mode} (Count:{male_max})")
```

```python
#Empty list
ewallet_sales = []
cash_sales = []
card_sales = []

#Sales by payment
for s,p in totals2:
    if p == "Ewallet":
        ewallet_sales.append(s)
    elif p == "Cash":
        cash_sales.append(s)
    elif p == "Credit card":
        card_sales.append(s)
#Totals
total_ewallet = sum (ewallet_sales)
total_cash = sum (cash_sales)
total_card = sum (card_sales)

#Max
highest_totals = max(total_ewallet, total_cash, total_card)

#Payment vs total
if highest_totals == total_ewallet:
    top_method = "Ewallet"
elif highest_totals == total_ewallet:
    top_method = "Cash"
else:
    top_method = "Credit card"

print (f"Payment method ({top_method}) tops with total sales of {highest_totals}")
```
